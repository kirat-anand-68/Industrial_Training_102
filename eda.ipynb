```python
plt.figure(figsize=(8,5))
sns.scatterplot(x='ProductCost', y='ProductPrice', data=pdt, color='teal')
plt.title("Product Cost vs Product Price")
plt.xlabel("Product Cost")
plt.ylabel("Product Price")
plt.show()
```

Category-based Analysis

```python
plt.figure(figsize=(10,5))
sns.boxplot(x='ProductColor', y='ProductPrice', data=pdt)
plt.title("Product Price Distribution by Color")
plt.xticks(rotation=45)
plt.show()
```

Profit Calculation

```python
pdt['Profit'] = pdt['ProductPrice'] - pdt['ProductCost']
print("\n💰 Profit Added Column:")
print(pdt[['ProductName', 'ProductCost', 'ProductPrice', 'Profit']].head())
```

Top Profitable Products 

```python
plt.figure(figsize=(8,5))
sns.histplot(pdt['Profit'], bins=20, kde=True, color='orange')
plt.title("Profit Distribution")
plt.show()
```

```python
top_profit = pdt.sort_values(by='Profit', ascending=False).head(10)
print("\n🏆 Top 10 Profitable Products:")
print(top_profit[['ProductName', 'ProductCost', 'ProductPrice', 'Profit']])
```

```python
plt.figure(figsize=(10,5))
sns.barplot(x='ProductName', y='Profit', data=top_profit, palette='viridis')
plt.title("Top 10 Profitable Products")
plt.xticks(rotation=45, ha='right')
plt.show()
```
